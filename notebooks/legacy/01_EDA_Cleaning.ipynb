```python


# 1. Imports
import pandas as pd
import numpy as np
import re
import os

# 2. Load Data
data_path = '/Users/ronitgandhi/Portfolio Projects/job_postings_insights/data/postings.csv'  # adjust path if needed

df = pd.read_csv(data_path)

# 3. Quick EDA
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nSample rows:")
display(df.sample(5))

# 4. Important Columns to Focus
important_cols = [
    'job_id', 'company_name', 'title', 'description', 'location', 
    'min_salary', 'max_salary', 'pay_period', 'remote_allowed', 
    'work_type', 'skills_desc'
]

df = df[important_cols]

# 5. Check Missing Values
print("\nMissing values per column:")
print(df.isnull().sum())

# 6. Basic Cleaning of Description
def clean_text(text):
    if pd.isnull(text):
        return ""
    # Remove multiple newlines, tabs
    text = re.sub(r'\n+', ' ', text)
    text = re.sub(r'\t+', ' ', text)
    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

df['description'] = df['description'].apply(clean_text)

# 7. Drop rows where description is too short (optional)
df = df[df['description'].str.len() > 50]

# 8. Save Cleaned Version
os.makedirs('data', exist_ok=True)
df.to_csv('data/cleaned_job_postings.csv', index=False)

print("\n✅ Cleaned data saved to data/cleaned_job_postings.csv")
```
